```python
from pathlib import Path
import time
import requests
import pandas as pd

# Folder for the original API downloads
output_folder = Path("berlin_air_quality_raw")
output_folder.mkdir(exist_ok=True)

# Keep only active stations
stations_df = pd.DataFrame(stations_data)
active_stations = stations_df[stations_df["active"] == True].copy()

# May to October, for both years
months = [
    (2025, 5), (2025, 6), (2025, 7),
    (2025, 8), (2025, 9), (2025, 10),
    (2026, 5), (2026, 6), (2026, 7),
    (2026, 8), (2026, 9), (2026, 10)
]

session = requests.Session()
errors = []
downloaded_files = []

for _, station in active_stations.iterrows():
    station_code = station["code"]
    station_name = station["name"]

    station_records = []

    for year, month in months:
        start = f"01.{month:02d}.{year} 00:00"

        # Last calendar day of the requested month
        last_day = pd.Period(f"{year}-{month:02d}").days_in_month
        end = f"{last_day:02d}.{month:02d}.{year} 23:00"

        url = f"https://luftdaten.berlin.de/api/stations/{station_code}/data"

        params = {
            "period": "1h",
            "timespan": "custom",
            "start": start,
            "end": end
        }

        try:
            response = session.get(url, params=params, timeout=60)
            response.raise_for_status()
            records = response.json()

            if not isinstance(records, list):
                raise ValueError("API response was not a list")

            station_records.extend(records)

            print(
                f"{station_code} | {year}-{month:02d} | "
                f"{len(records)} records"
            )

        except Exception as error:
            errors.append({
                "station": station_code,
                "year": year,
                "month": month,
                "error": str(error)
            })
            print(f"ERROR: {station_code} | {year}-{month:02d} | {error}")

        time.sleep(0.2)

    if station_records:
        station_df = pd.DataFrame(station_records)

        station_df["datetime"] = pd.to_datetime(
            station_df["datetime"], 
            errors="coerce",
            utc=True
        )

        station_df = station_df.sort_values(
            ["datetime", "core"]
        )

        filename = output_folder / f"{station_code}_air_quality.csv"
        station_df.to_csv(filename, index=False)

        downloaded_files.append({
            "station": station_name,
            "code": station_code,
            "records": len(station_df),
            "file": str(filename)
        })

# Summarize the results
download_summary = pd.DataFrame(downloaded_files)
error_summary = pd.DataFrame(errors)

print("\nDownload summary:")
display(download_summary)

print("\nErrors:")
display(error_summary)
```

```python
from pathlib import Path
import pandas as pd

file_path = Path("berlin_air_quality_raw/mc010_air_quality.csv")
wedding_df = pd.read_csv(file_path)

wedding_df["datetime"] = pd.to_datetime(
    wedding_df["datetime"],
    errors="coerce",
    utc=True
)

print("Rows:", len(wedding_df))
print("Date range:", wedding_df["datetime"].min(), "to", wedding_df["datetime"].max())
print("Stations:", wedding_df["station"].nunique())
print("Pollutants:", wedding_df["core"].unique())
print("Missing values:")
print(wedding_df[["datetime", "core", "value"]].isna().sum())
```

```python
quality_check = (
    wedding_df.groupby("core")
    .agg(
        total_records=("value", "size"),
        available_readings=("value", "count"),
        first_reading=("datetime", "min"),
        last_reading=("datetime", "max")
    )
)

quality_check["missing_values"] = (
    quality_check["total_records"]
    - quality_check["available_readings"]
)

quality_check["missing_pct"] = (
    quality_check["missing_values"]
    / quality_check["total_records"] * 100
).round(2)

display(quality_check)
```

```python
# Convert timestamps to Berlin local time for calendar-month analysis
wedding_df["datetime_berlin"] = (
    wedding_df["datetime"].dt.tz_convert("Europe/Berlin")
)

# Keep only June, July and August of 2025 and 2026
summer_df = wedding_df[
    (wedding_df["datetime_berlin"].dt.year.isin([2025, 2026]))
    & (wedding_df["datetime_berlin"].dt.month.isin([6, 7, 8]))
].copy()

# Count available readings for each pollutant by year and month
monthly_coverage = (
    summer_df
    .groupby([
        summer_df["datetime_berlin"].dt.year.rename("year"),
        summer_df["datetime_berlin"].dt.month.rename("month"),
        "core"
    ])["value"]
    .agg(
        total_records="size",
        available_readings="count"
    )
    .reset_index()
)

monthly_coverage["missing_pct"] = (
    (monthly_coverage["total_records"]
     - monthly_coverage["available_readings"])
    / monthly_coverage["total_records"] * 100
).round(2)

display(monthly_coverage)
```

```python
# Focus on July 2025 EBC readings
ebc_july = summer_df[
    (summer_df["core"] == "ebc")
    & (summer_df["datetime_berlin"].dt.year == 2025)
    & (summer_df["datetime_berlin"].dt.month == 7)
].copy()

# Summarize available readings by day
ebc_july["day"] = ebc_july["datetime_berlin"].dt.date

daily_ebc = (
    ebc_july.groupby("day")["value"]
    .agg(total_records="size", available_readings="count")
)

daily_ebc["missing_readings"] = (
    daily_ebc["total_records"] - daily_ebc["available_readings"]
)

display(daily_ebc)
```

```python
from pathlib import Path
import pandas as pd

folder = Path("berlin_air_quality_raw")
files = list(folder.glob("*_air_quality.csv"))

print("Number of station files:", len(files))

# Load and combine the station files
all_stations = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True
)

# Parse timestamps consistently
all_stations["datetime"] = pd.to_datetime(
    all_stations["datetime"],
    errors="coerce",
    utc=True
)

# Convert to Berlin local time
all_stations["datetime_berlin"] = (
    all_stations["datetime"].dt.tz_convert("Europe/Berlin")
)

# Keep June, July and August of 2025 and 2026
summer = all_stations[
    (all_stations["datetime_berlin"].dt.year.isin([2025, 2026]))
    & (all_stations["datetime_berlin"].dt.month.isin([6, 7, 8]))
].copy()

# Summarize the data available per station, year and pollutant
coverage = (
    summer.groupby([
        "station",
        summer["datetime_berlin"].dt.year.rename("year"),
        "core"
    ])
    .agg(
        records=("value", "size"),
        available_readings=("value", "count"),
        first_reading=("datetime_berlin", "min"),
        last_reading=("datetime_berlin", "max")
    )
    .reset_index()
)

coverage["missing_pct"] = (
    (coverage["records"] - coverage["available_readings"])
    / coverage["records"] * 100
).round(2)

display(coverage.sort_values(["station", "year", "core"]))
```

```python
# Count distinct timestamps with non-missing values
actual = (
    summer.groupby([
        "station",
        summer["datetime_berlin"].dt.year.rename("year"),
        summer["datetime_berlin"].dt.month.rename("month"),
        "core"
    ])
    .agg(
        rows=("datetime_berlin", "size"),
        unique_hours=("datetime_berlin", "nunique"),
        available_readings=("value", "count")
    )
    .reset_index()
)

# Expected hours for June, July and August
expected_hours = {
    6: 720,
    7: 744,
    8: 744
}

actual["expected_hours"] = actual["month"].map(expected_hours)

# A pollutant can have multiple records for the same hour,
# so compare unique timestamps to expected hours
actual["hour_coverage_pct"] = (
    actual["unique_hours"] / actual["expected_hours"] * 100
).round(2)

actual["reading_coverage_pct"] = (
    actual["available_readings"] / actual["expected_hours"] * 100
).round(2)

display(
    actual.sort_values(
        ["station", "year", "month", "core"]
    )
)
```

```python
# Overall summary for each station and pollutant
coverage_summary = (
    actual.groupby(["station", "core"])
    .agg(
        months_checked=("month", "size"),
        min_hour_coverage=("hour_coverage_pct", "min"),
        min_reading_coverage=("reading_coverage_pct", "min"),
        avg_reading_coverage=("reading_coverage_pct", "mean")
    )
    .round(2)
    .reset_index()
)

display(
    coverage_summary.sort_values(
        ["min_reading_coverage", "station", "core"]
    )
)
```

```python
# Focus on the four primary pollutants
primary_pollutants = ["pm10", "pm2", "no2", "o3"]

primary_coverage = actual[
    actual["core"].isin(primary_pollutants)
].copy()

# Find the worst monthly reading coverage per station,
# pollutant and year
station_quality = (
    primary_coverage
    .groupby(["station", "year", "core"])
    .agg(
        worst_month_coverage=("reading_coverage_pct", "min"),
        average_month_coverage=("reading_coverage_pct", "mean")
    )
    .round(2)
    .reset_index()
)

# Reshape so that 2025 and 2026 can be compared side by side
quality_comparison = station_quality.pivot(
    index=["station", "core"],
    columns="year",
    values="worst_month_coverage"
).reset_index()

quality_comparison.columns.name = None
quality_comparison = quality_comparison.rename(
    columns={
        2025: "worst_month_2025",
        2026: "worst_month_2026"
    }
)

display(
    quality_comparison.sort_values(
        ["core", "worst_month_2025", "worst_month_2026"],
        ascending=[True, False, False]
    )
)
```

```python
# Require at least 98% coverage in every month of both summers
quality_comparison["passes_2025"] = (
    quality_comparison["worst_month_2025"] >= 98
)

quality_comparison["passes_2026"] = (
    quality_comparison["worst_month_2026"] >= 98
)

quality_comparison["passes_both_years"] = (
    quality_comparison["passes_2025"]
    & quality_comparison["passes_2026"]
)

# Show which primary pollutants pass at each station
station_shortlist = (
    quality_comparison
    .pivot(
        index="station",
        columns="core",
        values="passes_both_years"
    )
    .fillna(False)
)

display(station_shortlist)
```

```python
from datetime import date
import meteostat as ms
from pathlib import Path

# Define the stations we want to download
stations = {
    "10384": "Tempelhof",
    "10389": "Alexanderplatz",
    "10382": "Tegel",
    "ETNB0": "Reinickendorf",
    "10385": "Schonefeld",
    "10379": "Potsdam"
}

# Set the historical date range
start = date(2016, 1, 1)
end = date(2026, 10, 20)

# Create the output folder if it doesn't exist
output_folder = Path("berlin_weather_daily")
output_folder.mkdir(exist_ok=True)

# Try downloading each station separately
for station_id, station_name in stations.items():
    try:
        # Request daily weather data for this station
        weather = ms.daily(
            ms.Station(id=station_id),
            start,
            end
        ).fetch()

        # Check whether Meteostat returned usable data
        if weather is None:
            print(f"NO DATA: {station_name} ({station_id}) returned None")
            continue

        if weather.empty:
            print(f"EMPTY: {station_name} ({station_id}) returned no rows")
            continue

        # Add the station details to the dataset
        weather["station_id"] = station_id
        weather["station_name"] = station_name

        # Save the dataset as a CSV file
        file_path = output_folder / f"{station_id}_{station_name.lower()}.csv"
        weather.to_csv(file_path, index=True, index_label="date")

        # Report the number of downloaded rows
        print(f"SUCCESS: {station_name} — {len(weather)} rows saved")

    except Exception as e:
        # Show the station and error so we can troubleshoot it
        print(f"ERROR: {station_name} ({station_id}) — {e}")
```

```python
import pandas as pd

# Define the period we want to investigate
start_year = 2016
end_year = 2026

# Convert station operating dates into datetime values
stations_df["measuringStart"] = pd.to_datetime(
    stations_df["measuringStart"], utc=True, errors="coerce"
)
stations_df["measuringEnd"] = pd.to_datetime(
    stations_df["measuringEnd"], utc=True, errors="coerce"
)

# Keep stations whose operating periods overlap 2016–2026
period_start = pd.Timestamp(f"{start_year}-01-01", tz="UTC")
period_end = pd.Timestamp(f"{end_year}-12-31 23:59:59", tz="UTC")

historical_stations = stations_df[
    (stations_df["measuringStart"] <= period_end)
    & (
        stations_df["measuringEnd"].isna()
        | (stations_df["measuringEnd"] >= period_start)
    )
].copy()

# Identify stations with hourly measurements for key pollutants
pollutants = ["pm10_1h", "pm2_1h", "no2_1h", "o3_1h", "no_1h", "nox_1h", "ebc_1h"]

def available_hourly_pollutants(components):
    # Return the relevant hourly pollutants listed for each station
    if not isinstance(components, list):
        return []
    return [pollutant for pollutant in pollutants if pollutant in components]

historical_stations["hourly_pollutants"] = historical_stations[
    "components"
].apply(available_hourly_pollutants)

# Display a compact list of potentially useful stations
station_overview = historical_stations[
    ["name", "code", "active", "measuringStart", "measuringEnd", "hourly_pollutants"]
].copy()

display(station_overview)
print("Stations overlapping 2016–2026:", len(station_overview))
```

```python
import requests
import pandas as pd

# Set the station and test periods
station_code = "mc010"
test_periods = {
    "January 2016": ("2016-01-01", "2016-01-31"),
    "October 2026": ("2026-10-01", "2026-10-27")
}

# Define the pollutants we want to test
pollutants_to_test = ["pm10", "pm2", "no2", "o3"]

# Test each period separately
for period_name, (start_date, end_date) in test_periods.items():
    for pollutant in pollutants_to_test:
        # Build the API request for one station, pollutant and period
        url = f"https://luftdaten.berlin.de/api/stations/{station_code}/data"
        params = {
            "component": pollutant,
            "period": "1h",
            "timespan": "custom",
            "start": start_date,
            "end": end_date
        }

        # Request the data from the API
        response = requests.get(url, params=params, timeout=30)
        print(f"\n{period_name} | {pollutant} | HTTP {response.status_code}")

        # Display the returned data structure
        if response.ok:
            result = response.json()

            if isinstance(result, dict):
                print("Response keys:", list(result.keys()))
                print("Sample:", str(result)[:400])
            elif isinstance(result, list):
                print("Number of records:", len(result))
                print("Sample:", result[:2])
            else:
                print("Response type:", type(result).__name__)
                print("Sample:", str(result)[:400])
        else:
            print(response.text[:300])
```

```python
# checking the actual date range and which pollutants are returned, rather than printing thousands of records.
import requests
import pandas as pd

# Test one station across the beginning and end of the target period
station_code = "mc010"
test_periods = [
    ("2016-01-01", "2016-01-31"),
    ("2026-10-01", "2026-10-27")
]

# Test hourly and daily measurements
for period in ["1h", "24h"]:
    for start_date, end_date in test_periods:
        # Request measurements for the selected period
        url = f"https://luftdaten.berlin.de/api/stations/{station_code}/data"
        params = {
            "period": period,
            "timespan": "custom",
            "start": start_date,
            "end": end_date
        }

        response = requests.get(url, params=params, timeout=60)
        print(f"\n{period}: {start_date} to {end_date}")

        # Summarise the returned records
        if response.ok:
            records = response.json()
            df = pd.DataFrame(records)

            if df.empty:
                print("No records returned")
                continue

            # Convert timestamps so we can inspect date coverage
            df["datetime"] = pd.to_datetime(df["datetime"], utc=True)

            print("Total records:", len(df))
            print("Unique timestamps:", df["datetime"].nunique())
            print("First timestamp:", df["datetime"].min())
            print("Last timestamp:", df["datetime"].max())

            # Count records by measured component
            display(df["component"].value_counts().to_frame("record_count"))
        else:
            print("HTTP error:", response.status_code, response.text[:200])
```

```python
import requests
import pandas as pd
from pathlib import Path

# Configure the pilot station and month
station_code = "mc010"
start_date = "2016-01-01"
end_date = "2016-01-31"

# Create a folder for the pilot downloads
output_folder = Path("berlin_air_quality_2016_pilot")
output_folder.mkdir(exist_ok=True)

# Download hourly and daily records for the selected month
for period in ["1h", "24h"]:
    # Build the API request
    url = f"https://luftdaten.berlin.de/api/stations/{station_code}/data"
    params = {
        "period": period,
        "timespan": "custom",
        "start": start_date,
        "end": end_date
    }

    # Request the data
    response = requests.get(url, params=params, timeout=60)
    response.raise_for_status()
    records = response.json()

    # Convert the records to a DataFrame
    df = pd.DataFrame(records)

    if df.empty:
        print(f"No data: {station_code}, {period}, {start_date}")
        continue

    # Save the raw timestamps and all returned components
    file_path = output_folder / f"{station_code}_{period}_2016-01.csv"
    df.to_csv(file_path, index=False)

    # Report what was saved
    print(f"Saved {len(df)} records to {file_path}")
```

```python
#getting hourly air quality data from 01.01.2025 to 20.10.2026
import requests
import pandas as pd
from pathlib import Path
from datetime import date
import time

# 1. Define the stations and API settings
stations = [
    "mc010", "mc027", "mc032", "mc042", "mc077",
    "mc085", "mc117", "mc124", "mc144", "mc145",
    "mc171", "mc174", "mc190", "mc221", "mc282"
]

base_url = "https://luftdaten.berlin.de/api/stations/{}/data"

# 2. Define the full date range (end date is exclusive)
start_date = date(2025, 1, 1)
end_date = date(2026, 10, 21)

# 3. Create a folder for the downloaded raw data
output_folder = Path("berlin_air_quality_2025_2026")
output_folder.mkdir(parents=True, exist_ok=True)

# 4. Generate monthly date ranges
months = pd.date_range(
    start=start_date,
    end=end_date,
    freq="MS"
)

# 5. Download each station month by month
for station in stations:
    station_records = []

    for month_start in months:
        # Set the end to the first day of the next month
        month_end = month_start + pd.offsets.MonthBegin(1)

        # Don't request dates beyond 21 October 2026
        month_end = min(month_end.date(), end_date)
        request_start = month_start.date()

        params = {
            "period": "1h",
            "timespan": "custom",
            "start": request_start.isoformat(),
            "end": month_end.isoformat()
        }

        url = base_url.format(station)

        try:
            # Request hourly data from the API
            response = requests.get(url, params=params, timeout=60)
            response.raise_for_status()
            records = response.json()

            # Add the returned records to this station's dataset
            if records:
                station_records.extend(records)

            print(
                f"{station}: {request_start} to {month_end} — "
                f"{len(records)} records"
            )

        except (requests.RequestException, ValueError) as error:
            print(
                f"ERROR for {station}, {request_start}: {error}"
            )

        # Pause briefly between requests
        time.sleep(0.2)

    # 6. Save one raw CSV per station
    if station_records:
        df = pd.DataFrame(station_records)

        # Parse timestamps consistently in UTC
        if "datetime" in df.columns:
            df["datetime"] = pd.to_datetime(
                df["datetime"], errors="coerce", utc=True
            )

        # Sort records by timestamp and component
        sort_columns = [
            col for col in ["datetime", "component"]
            if col in df.columns
        ]
        if sort_columns:
            df = df.sort_values(sort_columns)

        # Save the downloaded data without dropping any columns
        file_path = output_folder / f"{station}_air_quality_2025-01_to_2026-10-20.csv"
        df.to_csv(file_path, index=False)

        print(f"Saved {len(df)} records to {file_path}")

    else:
        print(f"No records downloaded for {station}")

print("Download process finished.")
```

```python
import requests
from bs4 import BeautifulSoup

# Open the oxygen graph page for station 139
url = "https://wasserportal.berlin.de/station.php"
params = {
    "anzeige": "g",
    "thema": "oog",
    "station": "139"
}

session = requests.Session()
response = session.get(url, params=params, timeout=60)

# Use the page's declared encoding
response.encoding = response.apparent_encoding

soup = BeautifulSoup(response.text, "html.parser")

print("Final URL:", response.url)
print("Status:", response.status_code)

# Print the complete form, including hidden fields and button attributes
for i, form in enumerate(soup.find_all("form"), start=1):
    print(f"\n{'=' * 60}")
    print(f"FORM {i}")
    print("Action:", form.get("action"))
    print("Method:", form.get("method"))
    print(form.prettify())

# Look for scripts that may explain how the download buttons work
print("\n" + "=" * 60)
print("RELEVANT JAVASCRIPT")

for script in soup.find_all("script"):
    script_text = script.string or script.get_text()

    if any(term in script_text.lower() for term in [
        "nstoffid", "input_1", "nymin", "nymax", "download", "csv"
    ]):
        print(script_text[:5000])
```
